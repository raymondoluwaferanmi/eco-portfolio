```python
import pandas as pd
import numpy as np
import requests
import plotly.express as px

print("Downloading World Bank data...")

# 1. Download Electricity Access (% of population)
url_elec = "https://api.worldbank.org/v2/country/all/indicator/EG.ELC.ACCS.ZS?format=json&date=2021&per_page=300"
elec_data = requests.get(url_elec).json()[1]
df_elec = pd.DataFrame(elec_data)[['countryiso3code', 'country', 'value']]
df_elec.columns = ['iso3c', 'country_name', 'electricity_access']

# 2. Download GDP per capita
url_gdp = "https://api.worldbank.org/v2/country/all/indicator/NY.GDP.PCAP.KD?format=json&date=2021&per_page=300"
gdp_data = requests.get(url_gdp).json()[1]
df_gdp = pd.DataFrame(gdp_data)[['countryiso3code', 'value']]
df_gdp.columns = ['iso3c', 'gdp_pc']

# 3. Merge them together
df = pd.merge(df_elec, df_gdp, on='iso3c')

# 4. Clean up (keep only actual countries, drop NAs)
df = df[df['iso3c'].str.len() == 3]
df = df.dropna()
df['gdp_pc'] = df['gdp_pc'].astype(float)
df['electricity_access'] = df['electricity_access'].astype(float)

print(f"✅ Data downloaded for {len(df)} countries!")
df.head()
```

```python
# Create an interactive scatter plot using Plotly
fig = px.scatter(
    df,
    x="gdp_pc",
    y="electricity_access",
    hover_name="country_name", # Shows country name when you hover!
    log_x=True, # GDP on a log scale
    size="gdp_pc", # Bubble size based on GDP
    color="electricity_access",
    color_continuous_scale="Viridis",
    title="SDG 7: Electricity Access vs. GDP per Capita (2021)",
    labels={
        "gdp_pc": "GDP per Capita (Log Scale, USD)",
        "electricity_access": "Access to Electricity (% of Population)"
    }
)

# Show the plot
fig.show()

# Save as a standalone HTML file
fig.write_html("sdg7_dashboard.html")
print("✅ Interactive dashboard saved as sdg7_dashboard.html!")
```
